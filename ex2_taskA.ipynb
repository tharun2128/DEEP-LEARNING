```python
print ("THARUN 24BAD126")
# TASK A
# Activation Functions Visualization

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Input values
x = np.linspace(-10, 10, 500)

# Sigmoid
sigmoid = 1 / (1 + np.exp(-x))

# Tanh
tanh = np.tanh(x)

# ReLU
relu = np.maximum(0, x)
```

```python
plt.figure(figsize=(12, 8))

# Sigmoid
plt.subplot(3, 1, 1)
plt.plot(x, sigmoid)
plt.title("Sigmoid Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

# Tanh
plt.subplot(3, 1, 2)
plt.plot(x, tanh)
plt.title("Tanh Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

# ReLU
plt.subplot(3, 1, 3)
plt.plot(x, relu)
plt.title("ReLU Activation Function")
plt.xlabel("Input")
plt.ylabel("Output")
plt.grid()

plt.tight_layout()
plt.show()
```

```python
print("ACTIVATION FUNCTION COMPARISON")
print("=" * 50)

print("Sigmoid")
print("Output Range : 0 to 1")
print("Saturation   : At very large positive/negative inputs")
print("Gradient     : Can become very small")
print()

print("Tanh")
print("Output Range : -1 to 1")
print("Saturation   : At very large positive/negative inputs")
print("Gradient     : Can become very small")
print()

print("ReLU")
print("Output Range : 0 to infinity")
print("Saturation   : Negative inputs produce 0")
print("Gradient     : 1 for positive inputs")
print()

print("ReLU is computationally efficient compared with Sigmoid and Tanh.")
```
