```python
# ============================================================
# EXPERIMENT 2 - TASK D
# DEEP LEARNING EXPERIMENT MANAGEMENT
#
# Name    : THARUN
# Roll No : 24BAD126
# ============================================================

import os
from google.colab import drive

print("=" * 60)
print("TASK D - DEEP LEARNING EXPERIMENT MANAGEMENT")
print("=" * 60)

# ------------------------------------------------------------
# 1. MOUNT GOOGLE DRIVE
# ------------------------------------------------------------

print("\n[1] Mounting Google Drive...")

drive.mount("/content/drive")

print("Google Drive mounted successfully!")


# ------------------------------------------------------------
# 2. CREATE EXPERIMENT FOLDER
# ------------------------------------------------------------

folder = "/content/drive/MyDrive/DL_Experiment_2"

os.makedirs(folder, exist_ok=True)
os.makedirs(folder + "/models", exist_ok=True)
os.makedirs(folder + "/results", exist_ok=True)

print("\n[2] Experiment folders created:")
print(folder)
print(folder + "/models")
print(folder + "/results")


# ------------------------------------------------------------
# 3. CREATE README FILE
# ------------------------------------------------------------

readme = """# EXPERIMENT 2
## Comparative Study of Activation Functions and Optimization Algorithms in Deep Learning

### Student Details

Name: THARUN

Roll No: 24BAD126

### Aim

To analyze the impact of different activation functions and optimization
algorithms on the performance of Artificial Neural Networks.

### Task A

Activation Functions:
- Sigmoid
- Tanh
- ReLU

### Task B

Performance comparison of:
- Sigmoid
- Tanh
- ReLU

### Task C

Optimization Algorithms:
- SGD
- Momentum
- RMSProp
- Adam

### Task D

Experiment management using:
- Google Colab
- Google Drive
- Git
- GitHub

### Dataset

MNIST handwritten digit dataset.

### Software

Python
TensorFlow
NumPy
Matplotlib
Google Colab
Git
GitHub
"""

readme_path = folder + "/README.md"

with open(readme_path, "w") as f:
    f.write(readme)

print("\n[3] README.md created successfully!")


# ------------------------------------------------------------
# 4. CREATE EXPERIMENT INFORMATION
# ------------------------------------------------------------

info = """EXPERIMENT 2

Name: THARUN
Roll No: 24BAD126

Activation Functions:
1. Sigmoid
2. Tanh
3. ReLU

Optimization Algorithms:
1. SGD
2. Momentum
3. RMSProp
4. Adam

Dataset:
MNIST

Platform:
Google Colab

Storage:
Google Drive

Version Control:
Git and GitHub
"""

info_path = folder + "/experiment_info.txt"

with open(info_path, "w") as f:
    f.write(info)

print("[4] Experiment information saved!")


# ------------------------------------------------------------
# 5. CREATE RESULTS FILE
# ------------------------------------------------------------

results = """EXPERIMENT 2 RESULTS

Task A:
Sigmoid, Tanh and ReLU activation functions were visualized.

Task B:
Sigmoid, Tanh and ReLU were compared using the same ANN architecture.

Task C:
SGD, Momentum, RMSProp and Adam were compared.

Task D:
Google Drive and Git/GitHub were used for experiment management.
"""

results_path = folder + "/results/results.txt"

with open(results_path, "w") as f:
    f.write(results)

print("[5] Results file created!")


# ------------------------------------------------------------
# 6. CREATE LOCAL GITHUB PROJECT FOLDER
# ------------------------------------------------------------

github_folder = "/content/DL_Experiment_2_GitHub"

os.makedirs(github_folder, exist_ok=True)

print("\n[6] GitHub project folder created:")
print(github_folder)


# ------------------------------------------------------------
# 7. COPY FILES TO GITHUB PROJECT
# ------------------------------------------------------------

os.system(
    f'cp "{readme_path}" "{github_folder}/README.md"'
)

os.system(
    f'cp "{info_path}" "{github_folder}/experiment_info.txt"'
)

print("[7] Files copied to GitHub project folder!")


# ------------------------------------------------------------
# 8. CREATE .gitignore
# ------------------------------------------------------------

gitignore = """__pycache__/
*.pyc
.ipynb_checkpoints/
"""

with open(github_folder + "/.gitignore", "w") as f:
    f.write(gitignore)

print("[8] .gitignore created!")


# ------------------------------------------------------------
# 9. INITIALIZE GIT
# ------------------------------------------------------------

print("\n[9] Initializing Git repository...")

os.system(
    f'cd "{github_folder}" && git init'
)

print("Git repository initialized!")


# ------------------------------------------------------------
# 10. CONFIGURE GIT
# ------------------------------------------------------------

os.system(
    f'cd "{github_folder}" && git config user.name "THARUN"'
)

os.system(
    f'cd "{github_folder}" && git config user.email "your_email@example.com"'
)

print("[10] Git configuration completed!")


# ------------------------------------------------------------
# 11. ADD FILES
# ------------------------------------------------------------

os.system(
    f'cd "{github_folder}" && git add .'
)

print("[11] Files added to Git!")


# ------------------------------------------------------------
# 12. CREATE COMMIT
# ------------------------------------------------------------

os.system(
    f'cd "{github_folder}" && git commit -m "Completed Experiment 2"'
)

print("[12] Git commit created!")


# ------------------------------------------------------------
# 13. SHOW GIT STATUS
# ------------------------------------------------------------

print("\n[13] GIT STATUS")
print("=" * 60)

os.system(
    f'cd "{github_folder}" && git status'
)


# ------------------------------------------------------------
# 14. SHOW COMMIT HISTORY
# ------------------------------------------------------------

print("\n[14] GIT COMMIT HISTORY")
print("=" * 60)

os.system(
    f'cd "{github_folder}" && git log --oneline'
)


# ------------------------------------------------------------
# 15. DISPLAY CREATED FILES
# ------------------------------------------------------------

print("\n[15] CREATED FILES")
print("=" * 60)

for root, dirs, files in os.walk(folder):
    for file in files:
        print(os.path.join(root, file))


# ------------------------------------------------------------
# FINAL OUTPUT
# ------------------------------------------------------------

print("\n")
print("=" * 60)
print("TASK D COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nGoogle Drive folder:")
print(folder)

print("\nGitHub project folder:")
print(github_folder)

print("\nCompleted:")
print("1. Google Drive mounted")
print("2. Experiment folders created")
print("3. README.md created")
print("4. Experiment information created")
print("5. Results file created")
print("6. Git repository initialized")
print("7. Files added to Git")
print("8. Git commit created")
print("9. Git status displayed")
print("10. Git commit history displayed")

print("\nNext step:")
print("Create a GitHub repository and upload the project.")
```
