```python
#THARUN
#24BAD126
# TASK B
# Performance Comparison

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

print("TensorFlow version:", tf.__version__)

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Normalize pixel values
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# Create validation set
x_val = x_train[-10000:]
y_val = y_train[-10000:]

x_train = x_train[:-10000]
y_train = y_train[:-10000]

print("Training data:", x_train.shape)
print("Validation data:", x_val.shape)
print("Testing data:", x_test.shape)
```

```python
def create_model(activation):

    model = tf.keras.Sequential([

        tf.keras.layers.Flatten(input_shape=(28, 28)),

        tf.keras.layers.Dense(128, activation=activation),

        tf.keras.layers.Dense(64, activation=activation),

        tf.keras.layers.Dense(10, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

```python
sigmoid_model = create_model("sigmoid")

sigmoid_history = sigmoid_model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=10,
    batch_size=128,
    verbose=1
)
```

```python
tanh_model = create_model("tanh")

tanh_history = tanh_model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=10,
    batch_size=128,
    verbose=1
)
```

```python
relu_model = create_model("relu")

relu_history = relu_model.fit(
    x_train,
    y_train,
    validation_data=(x_val, y_val),
    epochs=10,
    batch_size=128,
    verbose=1
)
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    sigmoid_history.history["accuracy"],
    label="Sigmoid"
)

plt.plot(
    tanh_history.history["accuracy"],
    label="Tanh"
)

plt.plot(
    relu_history.history["accuracy"],
    label="ReLU"
)

plt.title("Training Accuracy Comparison")
plt.xlabel("Epoch")
plt.ylabel("Training Accuracy")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    sigmoid_history.history["loss"],
    label="Sigmoid"
)

plt.plot(
    tanh_history.history["loss"],
    label="Tanh"
)

plt.plot(
    relu_history.history["loss"],
    label="ReLU"
)

plt.title("Training Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.plot(
    sigmoid_history.history["val_loss"],
    label="Sigmoid"
)

plt.plot(
    tanh_history.history["val_loss"],
    label="Tanh"
)

plt.plot(
    relu_history.history["val_loss"],
    label="ReLU"
)

plt.title("Validation Loss Comparison")
plt.xlabel("Epoch")
plt.ylabel("Validation Loss")
plt.legend()
plt.grid()

plt.show()
```

```python
print("=" * 70)
print("ACTIVATION FUNCTION PERFORMANCE")
print("=" * 70)

print("\nSIGMOID")
print("Training Accuracy :", round(sigmoid_history.history["accuracy"][-1], 4))
print("Validation Accuracy:", round(sigmoid_history.history["val_accuracy"][-1], 4))
print("Training Loss     :", round(sigmoid_history.history["loss"][-1], 4))
print("Validation Loss   :", round(sigmoid_history.history["val_loss"][-1], 4))

print("\nTANH")
print("Training Accuracy :", round(tanh_history.history["accuracy"][-1], 4))
print("Validation Accuracy:", round(tanh_history.history["val_accuracy"][-1], 4))
print("Training Loss     :", round(tanh_history.history["loss"][-1], 4))
print("Validation Loss   :", round(tanh_history.history["val_loss"][-1], 4))

print("\nRELU")
print("Training Accuracy :", round(relu_history.history["accuracy"][-1], 4))
print("Validation Accuracy:", round(relu_history.history["val_accuracy"][-1], 4))
print("Training Loss     :", round(relu_history.history["loss"][-1], 4))
print("Validation Loss   :", round(relu_history.history["val_loss"][-1], 4))
```

```python
print("=" * 50)
print("TEST ACCURACY")
print("=" * 50)

sigmoid_test_loss, sigmoid_test_acc = sigmoid_model.evaluate(
    x_test, y_test, verbose=0
)

tanh_test_loss, tanh_test_acc = tanh_model.evaluate(
    x_test, y_test, verbose=0
)

relu_test_loss, relu_test_acc = relu_model.evaluate(
    x_test, y_test, verbose=0
)

print("Sigmoid Test Accuracy:", round(sigmoid_test_acc, 4))
print("Tanh Test Accuracy   :", round(tanh_test_acc, 4))
print("ReLU Test Accuracy   :", round(relu_test_acc, 4))
```
